#### 1. Retrieve a list of the top 5 most popular books based on the number of times they have been ordered. Also order by author name in descending order

````python
#Run this cell to see the expected output
#do not modify this cell
print('This is the expected output')
q1
````

```
%%sql q1_result <<


````python
%%sql q1_result <<
Select b.title, a.name as author_name, Count(oi.book_id) as order_count
from books as b
Join authors as a
on b.author_id = a.author_id
Join order_items as oi
On b.book_id = oi.book_id
GROUP BY b.title
ORDER BY order_count DESC, author_name DESC
LIMIT 5;
````

````python
%%sql q2_result <<
Select a.name as author_name,
sum(b.price * oi.quantity) as total_revenue
FROM order_items oi
Join books b on oi.book_id = b.book_id
Join authors a ON b.author_id =a.author_id
GROUP BY author_name, a.author_id
Order by total_revenue DESC;
````

````python
%%sql q3_result <<
SELECT c.name as customer_name, count(o.order_id) as order_count
FROM customers c
JOIN orders o ON  c.customer_id = o.customer_id
GROUP BY c.name
ORDER BY order_count DESC, customer_name DESC
LIMIT 5;
````

````python
%%sql q5_result <<
Select r.review_id,
b.title as book_title,
a.name as author_name,
r.rating,
r.review_text
FROM reviews r
JOIN books b ON r.book_id =b.book_id
JOIN authors a ON b.author_id =a.author_id
ORDER BY r.review_id ASC
LIMIT 5;
````
